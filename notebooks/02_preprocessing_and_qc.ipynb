# 02 — Preprocessing and QC

Walkthrough of preprocessing for a single subject: bandpass filter, epoching, AutoReject.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

import matplotlib.pyplot as plt

from config import SAMPLING_RATE, TARGET_CHANNELS
from util.io import raw_eeg_path, read_eeg_data
from util.preprocessing import preprocess_EEG

%matplotlib inline

DATASET_ID = 2
SUBJECT_NUM = 1
```

```python
path = raw_eeg_path(DATASET_ID, SUBJECT_NUM)
raw = read_eeg_data(path, sfreq=SAMPLING_RATE)
clean_eeg, epochs = preprocess_EEG(raw, freq_filter=True, AR=True, fle=True)
print(f"Epochs: {len(epochs)}, shape per epoch: {epochs.get_data().shape[1:]}")
```

```python
psd = clean_eeg.compute_psd(method="welch", fmin=1, fmax=40, picks=TARGET_CHANNELS)
psd.plot()
plt.title(f"PSD — dataset {DATASET_ID}, subject {SUBJECT_NUM:03d}")
plt.show()
```
