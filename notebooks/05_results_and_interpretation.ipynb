```python
import json
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

import joblib
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import ConfusionMatrixDisplay

from config import RESULTS_DIR

%matplotlib inline
```

```python
metrics_path = Path(RESULTS_DIR) / "metrics.json"
if not metrics_path.exists():
    metrics_path = Path(RESULTS_DIR) / "metrics_xgboost.json"

with open(metrics_path) as f:
    metrics = json.load(f)
metrics
```

```python
artifact = joblib.load("classifier_models/saved_models/xgboost_eeg_classifier.joblib")
if isinstance(artifact, dict) and "feature_importances" in metrics:
    names = list(metrics["feature_importances"].keys())
    values = list(metrics["feature_importances"].values())
    plt.barh(names[::-1], values[::-1])
    plt.xlabel("Importance")
    plt.title("XGBoost top features")
    plt.tight_layout()
```

## Limitations

- Subject-level split reduces leakage but limits test set size with 88 subjects.
- Class imbalance (A/F/C) may bias recall for minority groups.
- Features are hand-crafted biomarkers, not end-to-end deep learning on raw waveforms.
- Results depend on preprocessing choices (bandpass, AutoReject thresholds).
