# 01 — Dataset Overview

Explores participant metadata from the Miltiadous et al. dementia EEG dataset.

**Citation:** [DOI 10.3390/data8060095](https://doi.org/10.3390/data8060095)

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

import matplotlib.pyplot as plt
import pandas as pd

from config import DATASETS
from util.io import get_participant_data

%matplotlib inline
```

```python
frames = []
for i, dataset_id in enumerate(DATASETS):
    df = get_participant_data()[i].copy()
    df["dataset_id"] = dataset_id
    frames.append(df)

participants = pd.concat(frames, ignore_index=True)
participants.head()
```

```python
print(f"Total participants: {len(participants)}")
print(participants["Group"].value_counts())
participants.groupby("Group")[["Age", "MMSE"]].agg(["mean", "std"])
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
participants["Group"].value_counts().plot(kind="bar", ax=axes[0], title="Class balance")
participants.boxplot(column="MMSE", by="Group", ax=axes[1])
plt.suptitle("")
plt.tight_layout()
```
